```python
import datetime

expenses = []

#ADD expense
def add_expense():
    print ("\n=====ADD EXPENSE=====")
    category = input("Enter the expense category(shopping): ")
    while True:
        try:
            amount_str = input("Enter the expense amount: ₹ ")
            amount = float(amount_str.replace(',', '')) # Remove commas for proper conversion
            if amount <= 0:
                print ("Invalid amount. Please enter a positive number.")
            else:
                break
        except ValueError:
            print ("Invalid input. Please enter a valid numeric amount (e.g., 1000 or 1000.50).")

    date = datetime.datetime.now().strftime("%d-%m-%Y %H:%M:%S")
    expenses.append({"category": category, "amount": amount, "date": date})
    print ("Expense added successfully!")

# View all expenses
def view_expenses():
    print ("\n=====EXPENSES=====")
    if not expenses:
        print ("No expenses found.")
        return


    for i, expense in enumerate (expenses,start=1):
      print (f"{i}. {expense['category']} - ₹{expense['amount']} - {expense['date']}")



# Calculate  total expense
def calculate_total_expenses():
    print ("\n=====TOTAL EXPENSES=====")

    total=0
    for expense in expenses:
        total += expense["amount"]
    print (f"Total expenses: ₹{total}")



# Category - wise expense
def category_wise_expenses():
    print ("\n=====CATEGORY WISE EXPENSES=====")

    category = input("Enter category:")

    total = 0
    for expense in expenses:
        if expense["category"] == category:
            total += expense["amount"]
    print (f"Total expenses in {category}: ₹{total}")


# Delete expense
def delete_expense():
    print ("\n=====DELETE EXPENSE=====")

    if not expenses:
        print ("No expenses found.")
        return

    view_expenses()

    while True:
        try:
            number = int (input("\nEnter expense number to delete:"))
            break
        except ValueError:
            print("Invalid input. Please enter a number.")


    if number>=1 and number<=len(expenses):
        del expenses[number-1]
        print ("Expense deleted successfully!")
    else:
        print ("Invalid expense number.")


# Main menu
while True :
  print ("\n=====EXPENSE TRACKER=====")
  print ("1. Add Expense")
  print ("2. View Expenses")
  print ("3. Calculate Total Expenses")
  print ("4. Category-wise Expenses")
  print ("5. Delete Expense")
  print ("6. Exit")

  choice = input("\nEnter your choice: ")
  if choice == "1":
    add_expense()
  elif choice == "2":
    view_expenses()
  elif choice == "3":
    calculate_total_expenses()
  elif choice == "4":
    category_wise_expenses ()
  elif choice =="5":
    delete_expense()
  elif choice =="6":
    print ("Thank you for using  the Expense Tracker. Goodbye!")
    break
  else:
    print ("Invalid choice. Please try again.")
```
